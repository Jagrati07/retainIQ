### Handling Missing Values in `TotalCharges`

**Observation:** 11 blank values were found in `TotalCharges`. On investigation, all 11 belonged to customers with `tenure = 0`.

**Action Taken:** Converted `TotalCharges` to numeric using `pd.to_numeric(errors="coerce")` and replaced the resulting `NaN` values with `0`.

**Reason:** Since these customers had zero recorded months of service, assigning zero total charges was a reasonable, domain-based assumption. Mean or median imputation could have introduced misleading charge values.

**Result:** All 11 missing values were handled, and all 7,043 customer records were retained.

```python
df['Churn']= df['Churn'].map({'Yes':1,'No':0})
print(df["Churn"].unique())
print(df["Churn"].isna().sum())
print(df['Churn'].value_counts())
```

### Encoding the Target Variable

**Observation:** The target column `Churn` originally contained two categorical values: `Yes` and `No`.

**Decision:** Converted `Yes` to `1` and `No` to `0`.

**Reasoning:** Machine learning classification algorithms require a numerical target representation. Using 1 for churn and 0 for non-churn makes the prediction target explicit and easy to interpret.

**Action Taken:** Applied dictionary mapping to convert the target values into integers.

```python
cat_columns = df.select_dtypes(include="str").columns
for col in cat_columns:
    print(f'\n{col}:')
    print(col, df[col].unique())

```

### Standardization of Service Categories

Observation: The dataset contains categories such as No, No phone service, and No internet service across service-related columns.

Decision: Preserved No phone service and No internet service as separate categories instead of merging them with No.

Reasoning: These categories represent different customer situations:

No indicates that the customer does not subscribe to a particular service or feature.

No phone service indicates that the customer does not have phone service.

No internet service indicates that the customer does not have internet service.

Merging these values could remove meaningful information about a customer's service usage and potentially reduce the model's ability to identify churn patterns.
